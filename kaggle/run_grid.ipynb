```python
# Cell 2: Configuration -- EDIT CELL_ID BEFORE RUNNING
# Each Kaggle session handles exactly one cell.  Pick the one you own.
# Available cells: c0, c1, c2, c3, c4   (c5 only after 4-expert winner known)

CELL_ID = 'c1'       # <--- CHANGE THIS (one session = one cell)
RESUME  = False      # <--- set True when restarting an interrupted session

# Read GH_PAT from Kaggle Secrets.
# Add it: notebook Settings -> Secrets -> Add New Secret, name=GH_PAT.
import os
try:
    from kaggle_secrets import UserSecretsClient
    _pat = UserSecretsClient().get_secret('GH_PAT')
    os.environ['GH_PAT'] = _pat
    print(f'GH_PAT loaded from Kaggle Secrets ({len(_pat)} chars)')
except Exception as _e:
    print(f'WARNING: could not load GH_PAT: {_e}')
    print("Set manually: os.environ['GH_PAT'] = 'ghp_...'")

print(f'CELL_ID={CELL_ID!r}  RESUME={RESUME}')
```

```python
# Cell 3: Clone the repo
import os, subprocess

REPO_URL  = 'https://github.com/jamesburton/dotLLM.git'
BRANCH    = 'issue/trackM-mote'
REPO_DIR  = '/kaggle/working/dotLLM'

if not os.path.isdir(os.path.join(REPO_DIR, '.git')):
    print(f'Cloning {REPO_URL} @ {BRANCH} ...')
    subprocess.run(
        ['git', 'clone', '--depth', '1', '--branch', BRANCH, REPO_URL, REPO_DIR],
        check=True,
    )
else:
    print('Repo present, pulling latest ...')
    subprocess.run(['git', '-C', REPO_DIR, 'pull', '--ff-only'], check=True)

print(f'Repo ready at {REPO_DIR}')
os.chdir(REPO_DIR)
print(f'cwd = {os.getcwd()}')
```

```python
# Cell 4: Run the assigned grid cell (train -> eval -> push)
import subprocess, sys

cmd = [
    sys.executable, 'kaggle/run_cell.py',
    '--cell-id', CELL_ID,
    '--manifest', 'kaggle/grid_manifest.json',
    '--repo-dir', REPO_DIR,
    '--results-remote', 'https://github.com/jamesburton/dotLLM',
]
if RESUME:
    cmd.append('--resume')

print('Running:', ' '.join(cmd))
subprocess.run(cmd, check=True)
print('Done.')
```
